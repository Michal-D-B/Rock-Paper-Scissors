```python
from random import randint

def main():

    rules()
    
    user_name = name()
    
    words = ["rock", "paper", "scissors"]

    rematch = 'yes'
    
    while rematch == 'yes':
        com, user  = game(3,words)
    
        end_score(com, user, user_name)
        
        rematch = end_game(user_name)
        
    print("^^^^ Thanks for playing ^^^^")



def game(n,words):
    
    com_count = 0
    user_count = 0
    thisdict = {
      "paper": "rock",
      "rock": "scissors",
      "scissors": "paper"
    }
    
    for _ in range(n):
        
        com = words[randint(0,2)]
        user = user_input(words)
        print(com)
        
        if user == com:
            print("--- Draw ---")     
        elif thisdict[user] == com:
            user_count += 1
            print("--- Win ---")
        else:
            com_count += 1
            print("---- Lose ----")
            
    return com_count, user_count
    


def user_input(word):
    
    user = ""
    while user not in word:
        
        user = input().strip().lower()
    return user


def end_score(com_count,user_count,name):
   
    if user_count == com_count:
        print('******   SUCK   ******')
    elif user_count > com_count:
        print('******   HERO   ******')
    else:
        print('******   LOSER   ******')
    print(f"|{name:^10}| Computer |\n|{user_count:^10}|{com_count:^10}|\n{'*' * 23}")


def end_game(name):
    
    user = ""
    while user not in ["yes","no"]:
        user = input(f"Rematch, {name} ? yes or no ").strip().lower()
    return user
        

def name():
    
    try:
        name = input("What's your name? ").strip().title().split()
        print(f"Hello {name[0]}, let's play")
        return name[0]
        
    except IndexError:
        print("Hello Silent Ninja, let's play")
        return "Silent Ninja"

def rules():
    print("| ROCK | PAPER | SCISSOR |\n\nRules:\n1. Input your choice: 'rock', 'paper', or 'scissors'.\n2. The computer responds with a random counter-move.\n3. Play repeats automatically for a 3-round loop.\n4. The final game results are calculated and displayed.")


main()
```
